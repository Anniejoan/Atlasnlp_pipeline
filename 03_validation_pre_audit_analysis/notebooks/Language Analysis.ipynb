## CLDR-grounded naive expansion (population-based language–country plausibility filter)

Instead of naively cross-joining every language in a dataset with every country in that dataset (which creates nonsense pairs like German–Argentina whenever a dataset lists both), we filter the naive cross-product down to only the (language, country) pairs that [CLDR's `country_language_population.tsv`](https://github.com/unicode-org/cldr/blob/main/tools/cldr-code/src/main/resources/org/unicode/cldr/util/data/country_language_population.tsv) confirms — i.e. a language is only attributed to a country if CLDR records that language as spoken there at all (any % of population, no threshold).


```python
import unicodedata
import re

CLDR_TSV_PATH = "cldr_country_language_population.tsv"


def _norm(s):
    """Lowercase, strip accents/apostrophe variants/punctuation for robust matching."""
    if not isinstance(s, str):
        return ""
    s = unicodedata.normalize("NFKD", s)
    s = "".join(c for c in s if not unicodedata.combining(c))
    s = s.replace("\u2019", "'").replace("\u02bc", "'")
    return re.sub(r"[^a-z0-9]+", " ", s.lower()).strip()


cldr = pd.read_csv(
    CLDR_TSV_PATH, sep="\t", dtype=str, quotechar='"', engine="python",
    keep_default_na=False, na_values=[""],
)
cldr = cldr.rename(columns={"#CName": "CName"})
cldr = cldr[cldr["CCode"].notna() & (cldr["CCode"] != "")].copy()
cldr["CName"] = cldr["CName"].str.strip()
cldr["Language"] = cldr["Language"].str.strip()
cldr["country_norm"] = cldr["CName"].map(_norm)
cldr["lang_norm"] = cldr["Language"].map(_norm)

# CLDR confirms a (country, language) pair simply by the row's presence --
# "any percentage of population" means we don't threshold on LPopulation at all.
CLDR_VALID_PAIRS = set(zip(cldr["country_norm"], cldr["lang_norm"]))
CLDR_COUNTRY_NORMS = set(cldr["country_norm"])
CLDR_LANG_NORMS = set(cldr["lang_norm"])

# --- Country name aliases: worldbench content_country -> CLDR CName ---
COUNTRY_ALIASES = {
    "Antigua and Barbuda": "Antigua & Barbuda",
    "Bosnia and Herzegovina": "Bosnia & Herzegovina",
    "Cabo Verde": "Cape Verde",
    "Congo (Congo-Brazzaville)": "Congo - Brazzaville",
    "Cote d'Ivoire": "C\u00f4te d\u2019Ivoire",
    "Democratic Republic of the Congo": "Congo - Kinshasa",
    "Myanmar": "Myanmar (Burma)",
    "Palestine": "Palestinian Territories",
    "Saint Kitts and Nevis": "St. Kitts & Nevis",
    "Saint Lucia": "St. Lucia",
    "Saint Vincent and the Grenadines": "St. Vincent & Grenadines",
    "S\u00e3o Tom\u00e9 and Pr\u00edncipe": "S\u00e3o Tom\u00e9 & Pr\u00edncipe",
    "Trinidad and Tobago": "Trinidad & Tobago",
}

# --- Language aliases: worldbench language_cleaned -> CLDR Language name(s) ---
# Only added where the WorldBench label is a known alternate name for, or an
# umbrella covering, one or more specific CLDR-listed varieties. Everything
# else relies on normalized exact matching; languages with no CLDR entry at
# all get zero confirmed countries (see diagnostics below).
LANGUAGE_ALIASES = {
    "Amazigh": ["Central Atlas Tamazight", "Standard Moroccan Tamazight"],
    "Tamazight": ["Central Atlas Tamazight", "Standard Moroccan Tamazight"],
    "Balochi": ["Western Balochi"],
    "Cree": ["Northern East Cree", "Oji-Cree", "Plains Cree", "Swampy Cree", "Woods Cree"],
    "Dari": ["Parsi-Dari", "Zoroastrian Dari"],
    "Nahuatl": ["Central Huasteca Nahuatl", "Eastern Huasteca Nahuatl", "Western Huasteca Nahuatl"],
    "Frisian": ["Eastern Frisian", "Northern Frisian", "Saterland Frisian", "Western Frisian"],
    "Sorbian": ["Lower Sorbian", "Upper Sorbian"],
    "Ndebele": ["North Ndebele", "South Ndebele"],
    "Hmong": ["Hmong Daw", "Hmong Njua"],
    "Kikongo": ["Kongo"],
    "Kirundi": ["Rundi"],
    "Newar": ["Newari"],
    "Tagalog": ["Filipino"],
    "Ilokano": ["Iloko"],
    "Kapampangan": ["Pampanga"],
    "Luganda": ["Ganda"],
    "Ojibwe": ["Ojibwa", "Western Ojibway"],
    "Twi": ["Akan"],
    "Buryat": ["Buriat"],
    "Yakut": ["Sakha"],
    "Pulaar": ["Fulah"],
    "Sereer": ["Serer"],
    "Tuvan": ["Tuvinian"],
}
```

```python
def _country_to_cldr_norm(country):
    alias = COUNTRY_ALIASES.get(country, country)
    n = _norm(alias)
    return n if n in CLDR_COUNTRY_NORMS else None


def _language_to_cldr_norms(language):
    n = _norm(language)
    if n in CLDR_LANG_NORMS:
        return {n}
    if language in LANGUAGE_ALIASES:
        return {_norm(x) for x in LANGUAGE_ALIASES[language]}
    return set()


def cldr_confirms(country, language):
    cn = _country_to_cldr_norm(country)
    lns = _language_to_cldr_norms(language)
    if cn is None or not lns:
        return False
    return any((cn, ln) in CLDR_VALID_PAIRS for ln in lns)


tierC_pairs = tierC_exp[["content_country", "language_cleaned"]].drop_duplicates()
tierC_pairs["cldr_confirmed"] = tierC_pairs.apply(
    lambda r: cldr_confirms(r["content_country"], r["language_cleaned"]), axis=1
)

countries_per_lang_naive_blind = countries_per_lang_naive  # blind cross-product, kept for reference
countries_per_lang_naive = (
    tierC_pairs[tierC_pairs["cldr_confirmed"]]
    .groupby("language_cleaned")["content_country"]
    .nunique()
)
```

```python
# Diagnostics: how much of the blind cross-product survives CLDR grounding,
# and which WorldBench language labels have no CLDR population data at all
# (these get zero CLDR-confirmed countries, since we can't confirm any).
n_total_pairs = len(tierC_pairs)
n_confirmed = int(tierC_pairs["cldr_confirmed"].sum())
print(f"Blind naive (country, language) pairs: {n_total_pairs}")
print(f"CLDR-confirmed pairs: {n_confirmed} ({n_confirmed / n_total_pairs:.1%})")

wb_langs = tierC_pairs["language_cleaned"].unique()
no_cldr_data = sorted(l for l in wb_langs if not _language_to_cldr_norms(l))
print(f"\nLanguages with NO CLDR population entry at all ({len(no_cldr_data)}), "
      f"excluded from CLDR-grounded naive counts:")
print(no_cldr_data)
```

```python
comparison = pd.DataFrame({
    "Controlled": countries_per_lang,
    "Naive": countries_per_lang_naive,  # CLDR-grounded (see cells above)
    "Naive (blind cross-product)": countries_per_lang_naive_blind,
}).fillna(0)

comparison["Inflation"] = comparison["Naive"] - comparison["Controlled"]

comparison.sort_values("Naive", ascending=False).head(15)
```

```python
custom_palette = ["#2000B1", "#75BBFD"]
comparison.sort_values("Naive", ascending=False).head(15)[
    ["Controlled", "Naive"]
].plot(kind="bar", color = custom_palette, figsize=(10,4), fontsize=15)

plt.legend(fontsize=15)

plt.title("Language Spread Inflation (Naive vs Controlled)", fontsize=15)
plt.ylabel("Number of Countries", fontsize=15)
plt.xlabel("Top Inflated Langauges", fontsize=15)
plt.savefig("fig5_language_coverage_naive_vs_controlled.pdf", bbox_inches="tight")
plt.savefig("fig5_language_coverage_naive_vs_controlled.png", dpi=300, bbox_inches="tight")
plt.show()
```

The blind cross-product wildly overstates language–country coverage: French, for example, was naively linked to 161 countries versus 47 under the controlled expansion (~3.4x inflation), simply because any country listed anywhere alongside French in a multilingual, multi-country dataset got credited.

Grounding the naive expansion in CLDR population data collapses almost all of that apparent inflation: French drops from 161 (blind) to 46 CLDR-confirmed countries — essentially matching the controlled count of 47, because the CLDR filter discards countries where French isn't actually a population language (e.g. Jordan, China, Japan) while picking up genuinely Francophone countries the controlled tier's stricter dataset-selection rule happened to miss (e.g. Burkina Faso, C\u00f4te d'Ivoire, Rwanda, Togo).

This holds broadly: once every naive (language, country) pair must be confirmed by actual population data rather than co-occurrence in a dataset's metadata, most of the apparent "naive expansion inflation" turns out to be an artifact of blind cross-joining rather than a real pattern of overreporting — with the exception of languages CLDR itself undercounts for second-language use (see caveat above), which can make the grounded naive count fall *below* the controlled count.

Prior work and dataset documentation often treat multilingual datasets as contributing equally to all associated languages and regions. Even when countries/regions are included in the documentation for multilingual datasets, they are not cleanly mapped to the languages that represent them and readers are often left to map or guess the appropriate match themselves. We compare a controlled expansion (monolingual dataset to many countries, multilingual dataset to one country) against a naive expansion grounded in CLDR population data (assigning a language to a country only if that language is attested at any population share in that country), and show that even this more principled naive baseline can diverge substantially from the controlled counts — in both directions — underscoring that language metadata alone, however it is expanded, is an unreliable proxy for geographic coverage.
\n**Caveat:** CLDR's `country_language_population.tsv` documents languages with an established resident population (native speakers, historical minorities, etc.), not general second-language / lingua-franca use. This means it systematically *undercounts* countries for languages taught or used widely as a second language — e.g. it does not list English for Norway, Iceland, or Japan even though English is broadly spoken there, because no population/census figure for English speakers is on record for those countries. So the CLDR-grounded naive count is a lower bound on population-confirmed presence, not a full picture of functional language use.

## How many languages are actually inflated?

`comparison` covers every language that appears in either the controlled or naive expansion. A language counts as "inflated" if its naive count exceeds its controlled count.


```python
n_languages_total = len(comparison)
n_inflated_blind = int((comparison["Naive (blind cross-product)"] > comparison["Controlled"]).sum())
n_inflated_cldr = int((comparison["Naive"] > comparison["Controlled"]).sum())

print(f"Total languages compared: {n_languages_total}")
print(f"Inflated under blind naive expansion:  {n_inflated_blind} / {n_languages_total} "
      f"({n_inflated_blind / n_languages_total:.1%})")
print(f"Inflated under CLDR-grounded naive:    {n_inflated_cldr} / {n_languages_total} "
      f"({n_inflated_cldr / n_languages_total:.1%})")
```

### Figure: Top CLDR-inflated languages

Restricting to languages where the CLDR-grounded naive count actually exceeds the controlled count (real, population-confirmed inflation, not the blind-cross-product artifact).


```python
inflated_cldr = comparison[comparison["Naive"] > comparison["Controlled"]].copy()
inflated_cldr["Inflation"] = inflated_cldr["Naive"] - inflated_cldr["Controlled"]

top_inflated = inflated_cldr.sort_values("Inflation", ascending=False).head(15)
top_inflated[["Controlled", "Naive"]].rename(columns={"Naive": "CLDR-grounded"}).plot(
    kind="bar", color=custom_palette, figsize=(10, 4), fontsize=15
)

plt.legend(fontsize=15)
plt.title("Languages with CLDR-Grounded Coverage Inflation", fontsize=15)
plt.ylabel("Number of Countries", fontsize=15)
plt.xlabel("Language", fontsize=15)
plt.savefig("fig6_top_cldr_inflated_languages.pdf", bbox_inches="tight")
plt.savefig("fig6_top_cldr_inflated_languages.png", dpi=300, bbox_inches="tight")
plt.show()

top_inflated[["Controlled", "Naive", "Inflation"]]
```

```python
def filter_datasets(
    df,
    language=None,
    content_countries=None,
    task=None,
    multiple_choice=None,
    producer_countries=None
):
    out = df.copy()

    if language is not None:
        out = out[out["language"] == language]

    if content_countries is not None:
        out = out[out["content_country"].isin(content_countries)]

    task_col = "normalized_task_category" if "normalized_task_category" in out.columns else "Task Category"
    if task is not None and task_col in out.columns:
        out = out[out[task_col] == task]

    if multiple_choice is not None and "Is dataset in multiple choice format?" in out.columns:
        out = out.copy()
        out["is_multiple_choice"] = out["Is dataset in multiple choice format?"].apply(normalize_mcq)
        out = out[out["is_multiple_choice"] == multiple_choice]

    if producer_countries is not None and "producer_country" in out.columns:
        out = out[out["producer_country"].isin(producer_countries)]

    return out
```

## Country-Disaggregability Audit of Evaluation Datasets

Among language-labelled evaluation datasets, how many actually support a *meaningful* comparison across countries that share that language? A shared language label is not enough on its own -- for a language X to support cross-country comparison we need, at minimum:

1. an evaluation-style task (QA, closed-set classification/MCQ, or NLI) with an objective, automatically-scorable label;
2. explicit (not inferred) per-dataset country provenance;
3. at least two, preferably three or more, countries covered;
4. the same task and label definitions across those countries;
5. sufficient examples per country;
6. comparable data collection methodology and splits across countries.

**What we can check automatically from WorldBench metadata:** (1), (2), (3), and a coarse proxy for (4) -- whether the countries come from one coordinated multi-country dataset (same `Dataset name`, strong same-task/same-label signal) versus independently-created single-country datasets that merely share a task category (weak signal, needs manual comparability review).

**What we cannot check automatically:** (5) sufficient per-country sample size and (6) comparable collection/splits -- WorldBench's metadata has no instance-count or split-methodology field at all. These are left as an explicit manual-review flag on the shortlist below rather than silently assumed.

Base population: `tierA_exp` (the controlled, non-blind-cross-joined expansion used throughout this notebook), restricted to task types eligible for objective cross-country scoring.


```python
# Task types treated as "evaluation-style, objectively scorable": QA, NLI
# (closest taxonomy match to textual inference), and anything explicitly
# flagged multiple-choice/closed-set format regardless of task category.
EVAL_TASK_CATEGORIES = {
    "Question Answering",
    "Semantics: Lexical and Sentence Level, Textual Inference",  # closest available proxy for NLI
}

audit_base = tierA_exp[
    tierA_exp["language_cleaned"].notna()
    & (tierA_exp["language_excluded_from_country_inference"] == False)
].copy()

eligible = audit_base[
    audit_base["normalized_task_category"].isin(EVAL_TASK_CATEGORIES)
    | (audit_base["Is dataset in multiple choice format?"] == "Yes")
].copy()

# Explicit (not inferred, not "Not stated") per-dataset country provenance.
explicit = eligible[
    eligible["Country Attribution Method"].str.contains("Explicit", na=False)
].copy()

print(f"Controlled, language-tagged dataset rows: {len(audit_base)}")
print(f"  -> eligible task type (QA / NLI-proxy / MCQ): {len(eligible)} rows, "
      f"{eligible['language_cleaned'].nunique()} languages")
print(f"     -> explicit country provenance: {len(explicit)} rows "
      f"({len(explicit) / len(eligible):.1%} of eligible), "
      f"{explicit['language_cleaned'].nunique()} languages")
```

```python
per_lang_any_provenance = eligible.groupby("language_cleaned")["content_country"].nunique()
per_lang_explicit = explicit.groupby("language_cleaned")["content_country"].nunique()

print("Among eligible (QA/NLI/MCQ) datasets, regardless of provenance:")
print(f"  Languages spanning >= 2 countries: {(per_lang_any_provenance >= 2).sum()}")
print(f"  Languages spanning >= 3 countries: {(per_lang_any_provenance >= 3).sum()}")

print("\nAmong eligible datasets with EXPLICIT country provenance only "
      "(the criterion that actually supports a defensible comparison):")
print(f"  Languages spanning >= 2 countries: {(per_lang_explicit >= 2).sum()}")
print(f"  Languages spanning >= 3 countries: {(per_lang_explicit >= 3).sum()}")
```

```python
# Same-family signal: does a single coordinated dataset alone span multiple
# countries (strong evidence of shared task/label definitions), or does the
# language only reach multiple countries by pooling independently-created
# single-country datasets (weak signal -- needs manual comparability check)?
dataset_country_span = (
    explicit.groupby(["language_cleaned", "Dataset name"])["content_country"]
    .nunique()
    .reset_index(name="countries_in_this_dataset")
)

multi_country_datasets = dataset_country_span[dataset_country_span["countries_in_this_dataset"] >= 2]
print(f"Single datasets that alone span >=2 countries (strong comparability signal): "
      f"{len(multi_country_datasets)}")
multi_country_datasets.sort_values("countries_in_this_dataset", ascending=False).head(15)
```

```python
def _summarize_language(lang_df):
    countries = sorted(lang_df["content_country"].unique())
    datasets = sorted(lang_df["Dataset name"].dropna().unique())
    max_span = (
        lang_df.groupby("Dataset name")["content_country"].nunique().max()
        if len(datasets) else 0
    )
    return pd.Series({
        "n_countries": len(countries),
        "n_datasets": len(datasets),
        "max_countries_single_dataset_metadata": max_span,
        "has_metadata_multi_country_dataset": max_span >= 2,
        "countries": "; ".join(countries),
        "dataset_names_preview": "; ".join(datasets[:10]) + (" ..." if len(datasets) > 10 else ""),
    })


shortlist_all = (
    explicit.groupby("language_cleaned")[explicit.columns.tolist()]
    .apply(_summarize_language)
    .sort_values(["n_countries", "has_metadata_multi_country_dataset"], ascending=False)
)

# NOTE: n_datasets (and every other per-language dataset count downstream) is
# NOT additive across languages -- a dataset tagged to multiple languages is
# counted once under each of them. Summing n_datasets across the qualifying
# languages therefore exceeds the number of unique datasets; see the
# long-form provenance file below for true unique-dataset counts.
qualifying = shortlist_all[shortlist_all["n_countries"] >= 2].copy()
n_unique_datasets_qualifying = explicit[
    explicit["language_cleaned"].isin(qualifying.index)
]["Dataset name"].nunique()

print(f"{len(shortlist_all)} languages have >=1 eligible, explicit-provenance dataset.")
print(f"{len(qualifying)} qualify on n_countries >= 2 -- this 34-language table is the scope of "
      f"country_disaggregability_shortlist.csv for the rest of this section.")
print(f"n_datasets summed across these {len(qualifying)} languages = {qualifying['n_datasets'].sum()}, "
      f"vs {n_unique_datasets_qualifying} actually-unique dataset names (non-additive, see note above).")
qualifying
```

**Reading this audit:** the funnel above goes from ~18k language-tagged rows down to a much smaller set of languages that can support even a nominally fair cross-country comparison. Two caveats to carry into the writeup:

- The "explicit provenance" and "eligible task type" filters are the strict, defensible criteria; the "any provenance" numbers above them are shown only for contrast, not as a claim.

## Sample-size collection: HuggingFace-hosted datasets

Per your scope decision, we only pursue sample sizes where they're reliably and cheaply obtainable: datasets hosted on HuggingFace, via the [`datasets-server` size API](https://huggingface.co/docs/datasets-server). Of the 828 unique datasets across the 34 qualifying languages, 56 resolve to a HuggingFace dataset ID; 38 of those returned a row count (the rest are gated/private, renamed beyond our known-alias list, or use a non-Parquet-convertible loading script). Everything else in the shortlist remains explicitly `NOT VERIFIABLE` -- we do not guess.

Results were fetched once and cached to `hf_size_results_cache.json` for reproducibility (re-run `fetch_hf_dataset_ids` + the API loop below with `REFRESH_CACHE = True` to refresh).


```python
import re
import time
import json as _json
import urllib.request
import urllib.parse
import urllib.error

HF_SIZE_CACHE_PATH = "hf_size_results_cache.json"
REFRESH_CACHE = False  # set True to re-query the HF API instead of using the cached results


def extract_hf_id(url):
    """Pull an org/name (or legacy single-name) HF dataset id out of a Dataset Link URL."""
    if not isinstance(url, str) or "huggingface.co" not in url:
        return None
    path = url.split("huggingface.co", 1)[1]
    segs = [s for s in path.split("/") if s and "?" not in s]
    if not segs:
        return None
    if segs[0] in {"collections", "spaces", "docs", "models", "blog", "papers"}:
        return None
    if segs[0] == "datasets":
        segs = segs[1:]
    if len(segs) == 1 and segs[0]:
        return segs[0]
    if len(segs) == 2 and segs[0] and segs[1]:
        return f"{segs[0]}/{segs[1]}"
    return None


# A few legacy no-namespace dataset ids that have since moved under an org.
KNOWN_RENAMES = {
    "glue": "nyu-mll/glue",
    "lex_glue": "coastalcph/lex_glue",
}

# NOTE: 'qualifying' above is the per-LANGUAGE aggregate table (>=2 countries).
# Rebuild the per-ROW dataset table for the same qualifying languages here.
qualifying_rows = explicit[explicit["language_cleaned"].isin(qualifying.index)].copy()

qualifying_rows["hf_dataset_id"] = qualifying_rows["Dataset Link (URL)"].map(extract_hf_id)
qualifying_rows["hf_dataset_id"] = qualifying_rows["hf_dataset_id"].map(lambda x: KNOWN_RENAMES.get(x, x))
unique_hf_ids = sorted(qualifying_rows["hf_dataset_id"].dropna().unique())
print(f"Qualifying datasets resolving to a HuggingFace id: {len(unique_hf_ids)} / "
      f"{qualifying_rows['Dataset name'].nunique()} unique datasets")
```

```python
def fetch_hf_size(hf_id, retries=2, timeout=15):
    url = f"https://datasets-server.huggingface.co/size?dataset={urllib.parse.quote(hf_id, safe='/')}"
    for attempt in range(retries + 1):
        try:
            req = urllib.request.Request(url, headers={"User-Agent": "worldbench-audit/1.0"})
            with urllib.request.urlopen(req, timeout=timeout) as resp:
                data = _json.loads(resp.read().decode("utf-8"))
            if "error" in data:
                return {"status": "error", "message": data["error"]}
            total = data.get("size", {}).get("dataset", {}).get("num_rows")
            return {"status": "ok", "num_rows": total}
        except urllib.error.HTTPError as e:
            if e.code == 404:
                return {"status": "not_found"}
            if attempt == retries:
                return {"status": "http_error", "code": e.code}
            time.sleep(1)
        except Exception as e:
            if attempt == retries:
                return {"status": "exception", "message": str(e)}
            time.sleep(1)


if REFRESH_CACHE:
    hf_size_results = {}
    for hf_id in unique_hf_ids:
        hf_size_results[hf_id] = fetch_hf_size(hf_id)
        time.sleep(0.3)
    _json.dump(hf_size_results, open(HF_SIZE_CACHE_PATH, "w"), indent=2)
else:
    hf_size_results = _json.load(open(HF_SIZE_CACHE_PATH))

n_resolved = sum(1 for r in hf_size_results.values() if r.get("status") == "ok")
print(f"Resolved row counts for {n_resolved} / {len(unique_hf_ids)} HuggingFace-hosted datasets.")
```

```python
# A repo's num_rows==0 from the size API means the viewer couldn't materialize
# rows (gated config, non-Parquet-convertible loading script, etc.) -- not a
# real size, so it shouldn't be cited as dataset-size evidence.
# QCRI/AraDiCE additionally resolves to num_rows=8, but that repo is a stub
# pointing to separate per-dialect child datasets (ArabicMMLU-lev,
# AraDiCE-BoolQ, ...); its own row count isn't AraDiCE's actual scope, so
# it's excluded too rather than cited as Arabic's minimum.
KNOWN_NONREPRESENTATIVE_HF_IDS = {"QCRI/AraDiCE"}

qualifying_rows["hf_num_rows"] = qualifying_rows["hf_dataset_id"].map(
    lambda hid: hf_size_results.get(hid, {}).get("num_rows") if hid else None
)
qualifying_rows["hf_size_valid"] = (
    qualifying_rows["hf_num_rows"].notna()
    & (qualifying_rows["hf_num_rows"] > 0)
    & ~qualifying_rows["hf_dataset_id"].isin(KNOWN_NONREPRESENTATIVE_HF_IDS)
)

dataset_sizes = (
    qualifying_rows[qualifying_rows["hf_dataset_id"].notna()]
    .groupby(["language_cleaned", "Dataset name", "hf_dataset_id"])
    .agg(
        countries=("content_country", lambda s: "; ".join(sorted(s.unique()))),
        hf_num_rows=("hf_num_rows", "first"),
        hf_size_valid=("hf_size_valid", "first"),
    )
    .reset_index()
    .sort_values(["language_cleaned", "hf_num_rows"], na_position="last")
)
dataset_sizes.to_csv("country_disaggregability_hf_dataset_sizes.csv", index=False)

n_ids_resolved = dataset_sizes["hf_dataset_id"].nunique()
n_ok_status = sum(1 for r in hf_size_results.values() if r.get("status") == "ok")
n_valid = dataset_sizes.loc[dataset_sizes["hf_size_valid"], "hf_dataset_id"].nunique()
print(f"{n_ids_resolved} unique HF dataset ids resolved from Dataset Link URLs across the 34 languages; "
      f"{len(unique_hf_ids) - n_ids_resolved} of the {len(unique_hf_ids)} extracted ids didn't appear in any "
      f"qualifying-language row after grouping (should be 0).")
print(f"{n_ok_status} got a non-error API response (status=ok). Of those, {n_valid} have a valid "
      f"(nonzero, non-artifact) total row count usable as size evidence; "
      f"{n_ok_status - n_valid} returned zero rows or are the known-non-representative AraDiCE stub; "
      f"{len(unique_hf_ids) - n_ok_status} failed to resolve at all (404 / gated / server error).")
dataset_sizes[dataset_sizes["hf_size_valid"]].sort_values("hf_num_rows")
```

```python
hf_valid_stats = (
    dataset_sizes[dataset_sizes["hf_size_valid"]]
    .groupby("language_cleaned")
    .agg(
        n_hf_valid_size_counts=("hf_dataset_id", "nunique"),
        min_hf_rows=("hf_num_rows", "min"),
        median_hf_rows=("hf_num_rows", "median"),
        max_hf_rows=("hf_num_rows", "max"),
    )
)
n_hf_ids_resolved_per_lang = (
    dataset_sizes.groupby("language_cleaned")["hf_dataset_id"].nunique().rename("n_hf_ids_resolved")
)

# IMPORTANT: joins onto `qualifying` (the 34-language table), not the full
# 89-language `shortlist_all` -- country_disaggregability_shortlist.csv stays
# scoped to the 34 qualifying languages throughout this section.
qualifying = qualifying.join(hf_valid_stats).join(n_hf_ids_resolved_per_lang)
qualifying["n_hf_ids_resolved"] = qualifying["n_hf_ids_resolved"].fillna(0).astype(int)
qualifying["n_hf_valid_size_counts"] = qualifying["n_hf_valid_size_counts"].fillna(0).astype(int)
qualifying["n_hf_size_unavailable"] = qualifying["n_hf_ids_resolved"] - qualifying["n_hf_valid_size_counts"]


def _hf_summary(r):
    if r["n_hf_valid_size_counts"] == 0:
        return "NOT VERIFIABLE (none of the candidate datasets resolved to a usable HF size record)"
    return (
        f"{int(r['n_hf_valid_size_counts'])}/{int(r['n_datasets'])} datasets have a valid HF total-size record "
        f"(rows: min={r['min_hf_rows']:,.0f}, median={r['median_hf_rows']:,.1f}, max={r['max_hf_rows']:,.0f}). "
        f"This is a TOTAL dataset size (all countries/splits pooled), not a per-country count -- "
        f"see verified_per_country_range for actual per-country evidence where available."
    )


qualifying["hf_total_size_summary"] = qualifying.apply(_hf_summary, axis=1)

print("Per-language HF id/size counts are NOT additive across languages (a dataset tagged to multiple "
      "languages is counted once per language it's tagged under).")
print(f"Global (deduplicated) totals across the 34-language scope: "
      f"{dataset_sizes['hf_dataset_id'].nunique()} ids resolved, "
      f"{dataset_sizes.loc[dataset_sizes['hf_size_valid'],'hf_dataset_id'].nunique()} with a valid size count.")
qualifying[qualifying["n_hf_valid_size_counts"] > 0][
    ["n_countries", "n_datasets", "n_hf_ids_resolved", "n_hf_valid_size_counts",
     "min_hf_rows", "median_hf_rows", "max_hf_rows"]
].sort_values("n_hf_valid_size_counts", ascending=False)
```

**Reading these numbers:** a valid HF total-size record is the size of the *whole* dataset (all countries/splits pooled, since WorldBench doesn't track per-country row counts) -- treat it as an upper bound on any single country's sample within that dataset, not a per-country figure. Zero-row API responses and the QCRI/AraDiCE stub (a collection page, not the actual data) are excluded from these stats entirely rather than cited as if they were real sizes -- see the exclusion note in the cell above. This updates `country_disaggregability_shortlist.csv` in place -- the `hf_total_size_summary` column now reports the valid HF-confirmed counts where available instead of a blanket "NOT VERIFIABLE".


## Going beyond HF totals: dataset cards and paper tables

Two extra passes beyond the HF `size` API total, both hand-verified rather than trusted blindly:

1. **HF dataset-card configs.** The same `size` API call also returns a per-`config` breakdown when a dataset defines one config per split. Checked all 56 HF-linked datasets: only **ArabCulture** organizes its configs by country (the other 12 multi-config datasets split by task/subject/language instead, e.g. GLUE's `cola`/`mnli`/..., KazMMLU's subjects) -- so this recovers one genuine per-country table for free from data already fetched.
2. **Paper PDFs.** Scoped to the 21 datasets most worth the effort: the 6 "coordinated multi-country" datasets outside English/Arabic (small in number), all 5 Arabic ones (the language with the most concerning HF-confirmed sizes), and the top 10 English ones by country count. WebFetch's page-image summarization garbled every PDF table it was given -- direct text extraction via `pypdf` on the downloaded PDF worked far better and is what's used below. Every regex-matched "country name next to a number" candidate was then read in context and checked against its table caption; most were false positives (accuracy/score tables, shared-task participant-team counts, citation noise) and are excluded rather than reported as sample sizes. Full per-dataset verdicts (VERIFIED/REJECTED, with the reason) are in `DATASET_VERIFICATION_STATUS` below and in `country_disaggregability_dataset_assessments.csv`.

Of the 56 HF-linked datasets, 38 returned a non-error API response; 32 of those have a valid, non-artifact nonzero row count (5 returned 0 rows and QCRI/AraDiCE's 8-row stub is excluded as non-representative -- see the exclusion cell above); the remaining 18 failed to resolve at all.

### Long-form dataset provenance (no truncation)

`dataset_names` in the language-level shortlist is capped at 10 names for readability, which isn't enough for reproducible auditing on languages with dozens-to-hundreds of datasets (English alone has 553). `country_disaggregability_dataset_provenance.csv` gives one row per (language, dataset) pair -- 1,008 rows for the 34 qualifying languages, non-additive with the 828 unique-dataset count for the same reason as `n_datasets` above -- with the full metadata country list, HF id/size where resolved, and verification status.


```python
long_form = (
    qualifying_rows.groupby(["language_cleaned", "Dataset name"])
    .agg(
        countries_from_metadata=("content_country", lambda s: "; ".join(sorted(s.unique()))),
        n_countries_from_metadata=("content_country", "nunique"),
        hf_id=("hf_dataset_id", "first"),
        hf_total_rows=("hf_num_rows", "first"),
        hf_size_valid=("hf_size_valid", "first"),
    )
    .reset_index()
    .rename(columns={"language_cleaned": "language", "Dataset name": "dataset_name"})
)


def _row_verification(r):
    key = (r["language"], r["dataset_name"])
    if key in DATASET_VERIFICATION_STATUS:
        status, note = DATASET_VERIFICATION_STATUS[key]
        if status == "VERIFIED":
            source = "HF dataset configs" if r["dataset_name"] == "ArabCulture" else "paper table"
        else:
            source = "paper PDF (checked, not confirmed)"
        return pd.Series({
            "country_counts_verified": status == "VERIFIED",
            "verification_source": source,
            "verification_status": status,
            "notes": note,
        })
    return pd.Series({
        "country_counts_verified": False,
        "verification_source": None,
        "verification_status": "UNVERIFIED",
        "notes": None,
    })


long_form = pd.concat([long_form, long_form.apply(_row_verification, axis=1)], axis=1)
long_form.to_csv("country_disaggregability_dataset_provenance.csv", index=False)
print(f"{len(long_form)} (language, dataset) pairs across the 34 qualifying languages "
      f"({long_form['dataset_name'].nunique()} unique dataset names -- non-additive, "
      f"same caveat as n_datasets in the language-level shortlist).")
long_form[long_form["verification_status"] != "UNVERIFIED"]
```

**Reading the smallest-country-count numbers:** with only 17 examples (CULTURAL BENCH's Saudi Arabia), a binary-accuracy estimate has a worst-case ~95% margin of error around \u00b124 points; even at 59 examples (CULTURAL BENCH's China) it's roughly \u00b113 points. The precise claim is not that this makes comparison "unfair" -- it's that CULTURAL BENCH's small and uneven country samples don't support equally precise or reliable cross-country estimates. ArabCulture, by contrast, is relatively balanced (239-290 across all 13 countries). ArabicMMLU sits at the other extreme: 104-6,053, an over 55x range dominated by Jordan/Egypt/Palestine.

This is now reflected in `country_disaggregability_shortlist.csv` via `coordinated_dataset_verification_status`, `max_verified_disaggregable_countries`, and `verified_per_country_range` for Arabic and English; every other qualifying language still reads `UNVERIFIED` (never checked) or `REJECTED` (checked, no confirmed coordinated dataset -- French, Urdu, Bengali, Tigrinya, Russian), since paper-checking beyond this 21-dataset scope wasn't pursued. Per-dataset detail, including the comparability assessment, is in `country_disaggregability_dataset_assessments.csv`; full per-(language, dataset) provenance without truncation is in `country_disaggregability_dataset_provenance.csv`.
